```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'laptop_sales': [12, 15, 14, 18, 25, 30, 22],
    'phone_sales': [20, 22, 19, 24, 35, 40, 28],
    'accessories_sales': [5, 8, 7, 10, 15, 18, 12],
    'ad_spend_usd': [150, 180, 160, 210, 300, 350, 260],
    'avg_customer_age': [24, 28, 22, 35, 42, 38, 29]
}

df = pd.DataFrame(data)

df['total_sales'] = df['laptop_sales'] + df['phone_sales'] + df['accessories_sales']

df
```

```python
day_order = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
df['day'] = pd.Categorical(df['day'], categories=day_order, ordered=True)
```

```python
df.info()
```

**Line Chart:** Shows trends over continuous time

```python
plt.figure(figsize=(8, 4))
plt.plot(df['day'], df['laptop_sales'], marker='o', label='Laptops', color='steelblue')
plt.plot(df['day'], df['phone_sales'], marker='s', label='Phones', color='darkorange')

plt.title('Daily Sales Trend', fontweight='bold')
plt.xlabel('Day of Week')
plt.ylabel('Units Sold')
plt.grid(True, linestyle='--', alpha=0.5) # Set grid opacity to 50% (where 0.0 is completely transparent and 1.0 is fully solid)
plt.legend()
plt.show()
```

**Vertical Bar Chart:** Compares values across discrete categories

```python
plt.figure(figsize=(8, 4))
plt.bar(df['day'], df['total_sales'], color='teal', edgecolor='black')

plt.title('Total Daily Units Sold', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Total Units')
plt.show()
```

**Horizontal Bar Chart:** Compares categories with long labels or rankings

```python
plt.figure(figsize=(8, 4))
plt.barh(df['day'], df['ad_spend_usd'], color='coral', edgecolor='black')

plt.title('Daily Ad Spend ($)', fontweight='bold')
plt.xlabel('Ad Spend (USD)')
plt.ylabel('Day')
plt.show()
```

**Scatter Plot:** Shows relationship/correlation between two numeric variables

```python
plt.figure(figsize=(8, 4))
plt.scatter(df['ad_spend_usd'], df['total_sales'], color='purple', alpha=0.8)

plt.title('Ad Spend vs. Total Units Sold', fontweight='bold')
plt.xlabel('Ad Spend ($)')
plt.ylabel('Total Units Sold')
plt.grid(True, linestyle=':')
plt.show()
```

**Histogram:** Displays frequency distribution across continuous intervals

```python
plt.figure(figsize=(8, 4))
plt.hist(df['avg_customer_age'], bins=8, color='seagreen', edgecolor='white')

plt.title('Distribution of Avg Customer Age', fontweight='bold')
plt.xlabel('Age Range')
plt.ylabel('Frequency (Days)')
plt.show()
```

**Box Plot:** Summarizes statistical distribution (median, quartiles, outliers)

```python
plt.figure(figsize=(7, 4))
plt.boxplot([df['laptop_sales'], df['phone_sales'], df['accessories_sales']],
            tick_labels=['Laptops', 'Phones', 'Accessories'])

plt.title('Sales Volume Dispersion by Product', fontweight='bold')
plt.ylabel('Units Sold')
plt.show()
```

**Pie Chart:** Displays proportions of a whole (100%)

```python
category_totals = [df['laptop_sales'].sum(), df['phone_sales'].sum(), df['accessories_sales'].sum()]
categories = ['Laptops', 'Phones', 'Accessories']

plt.figure(figsize=(6, 6))
plt.pie(category_totals, labels=categories, autopct='%1.1f%%',
        colors=['r','g','b'], startangle=90)

plt.title('Weekly Sales Volume Share', fontweight='bold')
plt.show()
```

```python
category_totals = [df['laptop_sales'].sum(), df['phone_sales'].sum(), df['accessories_sales'].sum()]
categories = ['Laptops', 'Phones', 'Accessories']

plt.figure(figsize=(6, 6))
plt.pie(category_totals, labels=categories, autopct='%1.1f%%',
        colors=['r','g','b'], startangle=90, explode=(0.1,0,0))

plt.title('Weekly Sales Volume Share', fontweight='bold')
plt.show()
```

```python
category_totals = [df['laptop_sales'].sum(), df['phone_sales'].sum(), df['accessories_sales'].sum()]
categories = ['Laptops', 'Phones', 'Accessories']

plt.figure(figsize=(6, 6))
plt.pie(category_totals, labels=categories, autopct='%1.1f%%',
        colors=['r','g','b'], startangle=90)
plt.pie([1], radius=0.8, colors='w')

plt.title('Weekly Sales Volume Share', fontweight='bold')
plt.show()
```

**Stacked Bar Chart:** Compares category totals while showing sub-components

```python
plt.figure(figsize=(8, 4))
plt.bar(df['day'], df['laptop_sales'], label='Laptops', color='r')
plt.bar(df['day'], df['phone_sales'], bottom=df['laptop_sales'], label='Phones', color='g')
plt.bar(df['day'], df['accessories_sales'], bottom=df['laptop_sales'] + df['phone_sales'], label='Accessories', color='b')

plt.title('Daily Sales Composition', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Units Sold')
plt.legend()
plt.show()
```

**Clustered Bar Chart:** A Clustered (or Grouped) Bar Chart is used when you want to compare multiple sub-categories across main groups side-by-side.

```python
plt.figure(figsize=(10, 5))

# 1. Create numeric X positions and set bar width
x = np.arange(len(df['day']))
width = 0.25

# 2. Shift each product's bar along the X-axis
plt.bar(x - width, df['laptop_sales'], width=width, label='Laptops', color='r')
plt.bar(x,         df['phone_sales'],  width=width, label='Phones', color='g')
plt.bar(x + width, df['accessories_sales'], width=width, label='Accessories', color='b')

# 3. Replace numeric X positions with the Day string labels
plt.xticks(ticks=x, labels=df['day'])

plt.title('Daily Sales Comparison', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Units Sold')
plt.legend()
plt.show()
```

**Area Chart:** Emphasizes volume or magnitude over time

```python
plt.figure(figsize=(8, 4))
plt.stackplot(df['day'], df['laptop_sales'], df['phone_sales'], df['accessories_sales'],
              labels=['Laptops', 'Phones', 'Accessories'], alpha=0.7)

plt.title('Cumulative Weekly Sales Volume', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Units')
plt.legend(loc='upper left')
plt.show()
```

**Violin Plot:** Combines box plot stats with probability density shape

```python
plt.figure(figsize=(7, 4))
plt.violinplot([df['laptop_sales'], df['phone_sales']], showmedians=True)

plt.xticks([1, 2], ['Laptops', 'Phones'])
plt.title('Sales Density Spread (Violin Plot)', fontweight='bold')
plt.ylabel('Units Sold')
plt.show()
```

**Heatmap:** Displays data density or correlations via color intensity

```python
# Compute correlation matrix across numeric columns
corr = df[['laptop_sales', 'phone_sales', 'ad_spend_usd', 'avg_customer_age']].corr()

plt.figure(figsize=(6, 5))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar(label='Correlation Coefficient')
plt.xticks(range(len(corr)), corr.columns, rotation=90)
plt.yticks(range(len(corr)), corr.columns)
plt.title('Feature Correlation Heatmap', fontweight='bold')
plt.show()
```
