```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Trim extra spaces and standardize text casing (Title Case)
text_columns = ['Region', 'CustomerType', 'Category', 'Product', 'Returned']

for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
# Remove '$' signs from UnitPrice and convert to numbers

df['UnitPrice'] = df['UnitPrice'].astype(str).str.replace('$', '', regex=False)
# regex = False: Do not treat $ as a special pattern code. Just look for the literal dollar sign character $ in the text and remove it.


df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
# errors='coerce': If you run into text that cannot be converted to a number, do not crash! Just force (coerce) that value into NaN (Not a Number / Missing Value).
```

```python
# Remove '%' signs from Discount and convert to numbers
df['Discount'] = df['Discount'].astype(str).str.replace('%', '', regex=False)
df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')
```

```python
# Convert whole numbers like 10 into decimal values (0.10)
df.loc[df['Discount'] > 1, 'Discount'] = df['Discount'] / 100
```

```python
# Convert OrderDate into standard YYYY-MM-DD format
df['OrderDate'] = pd.to_datetime(df['OrderDate'], format='mixed', errors='coerce')
```

```python
df = df.dropna()
```

```python
df.info()
```

```python
for col in df.columns:
    print(f"--- Unique values in '{col}' ---")
    print(df[col].unique())
    print("\n")
```

```python
# Remove negative quantities, negative prices, and unrealistic price outliers
df = df[df['Quantity'] > 0]
df = df[df['UnitPrice'] > 0]
```

```python
# Replace string 'Nan' with real np.nan
df = df.replace('Nan', np.nan)

# Now dropna() will catch and remove them!
df = df.dropna()

# Reset the index so row numbers are sequential
df = df.reset_index(drop=True)
```

```python
df.info()
```

```python
# Statistical summary for key numerical and financial columns
important_num_cols = ['Quantity', 'UnitPrice', 'Discount']
df[important_num_cols].describe().round(2)
```

```python
# Summary for categorical text columns
important_cat_cols = ['Region', 'CustomerType', 'Category', 'Product']

df[important_cat_cols].describe()
```

```python
important_num_cols = ['Quantity', 'UnitPrice', 'Discount']

Q1 = df[important_num_cols].quantile(0.25)
Q3 = df[important_num_cols].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_percentages = ((df[important_num_cols] < lower_bound) | (df[important_num_cols] > upper_bound)).mean() * 100
print("Percentage of outliers per column:")
print(outlier_percentages.round(2).astype(str) + "%")

within_bounds = ((df[important_num_cols] >= lower_bound) & (df[important_num_cols] <= upper_bound)).all(axis=1)
print(f"\nTotal outlier rows found across all columns: {len(df) - within_bounds.sum()}")
```

```python
# Quantity: Median
qty_out = (df['Quantity'] < lower_bound['Quantity']) | (df['Quantity'] > upper_bound['Quantity'])
df.loc[qty_out, 'Quantity'] = df['Quantity'].median()

# UnitPrice: Mean
price_out = (df['UnitPrice'] < lower_bound['UnitPrice']) | (df['UnitPrice'] > upper_bound['UnitPrice'])
df.loc[price_out, 'UnitPrice'] = df['UnitPrice'].mean()

# Discount: Drop rows
disc_valid = (df['Discount'] >= lower_bound['Discount']) & (df['Discount'] <= upper_bound['Discount'])
df = df[disc_valid].reset_index(drop=True)
```

```python
# Gross Sales = Quantity * Unit Price
df['GrossSales'] = df['Quantity'] * df['UnitPrice']

# Discount Amount = Gross Sales * Discount Rate
df['DiscountAmount'] = df['GrossSales'] * df['Discount']

# Net Sales = Gross Sales - Discount Amount
df['NetSales'] = df['GrossSales'] - df['DiscountAmount']

# Calculate Cost of Goods Sold (COGS)
df['UnitCost'] = df['UnitPrice'] * 0.60
df['COGS'] = df['Quantity'] * df['UnitCost']

# Gross Profit = Net Sales - Cost of Goods Sold
df['GrossProfit'] = df['NetSales'] - df['COGS']

# Operating Expenses = 15% of Net Sales
df['OperatingExpenses'] = df['NetSales'] * 0.15

# Net Profit = Gross Profit - Operating Expenses
df['NetProfit'] = df['GrossProfit'] - df['OperatingExpenses']
```
