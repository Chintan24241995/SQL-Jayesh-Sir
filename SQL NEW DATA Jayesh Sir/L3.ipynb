```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('cleaned_sales_data.csv')
df.head()
```

```python
df.info()
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
```

```python
df.info()
```

**EDA**

```python
# Aggregate Net Sales, Cost, and Profit by Category
category_eda = df.groupby('Category')[['NetSales', 'COGS', 'GrossProfit', 'NetProfit']].sum()
category_eda = category_eda.sort_values(by='NetSales', ascending=False)

category_eda.round(2)
```

```python
# Analyze sales and profit across different geographic regions
region_eda = df.groupby('Region')[['NetSales', 'NetProfit', 'Quantity']].agg({
    'NetSales': 'sum',
    'NetProfit': 'sum',
    'Quantity': 'sum'
})

region_eda.round(2)
```

```python
# Identify the top 5 profit-generating products
top_profitable_products = df.groupby('Product')['NetProfit'].sum().nlargest(5).reset_index()

top_profitable_products.round(2)
```

```python
# Analyze average order value and total spend per customer segment
customer_eda = df.groupby('CustomerType').agg(
    TotalRevenue=('NetSales', 'sum'),
    AverageOrderValue=('NetSales', 'mean'),
    TotalOrders=('OrderID', 'count')
)

customer_eda.round(2)
```

```python
# Group data by year-month to track performance over time
df['Month'] = df['OrderDate'].dt.to_period('M')

monthly_eda = df.groupby('Month')[['NetSales', 'GrossProfit', 'NetProfit']].sum()

monthly_eda.round(2)
```

```python
# Compare profit margins across different discount levels
discount_eda = df.groupby('Discount').agg(
    TotalOrders=('OrderID', 'count'),
    TotalNetSales=('NetSales', 'sum'),
    TotalNetProfit=('NetProfit', 'sum')
)

discount_eda.round(2)
```

```python
# Calculate return counts and lost net sales per product category
return_eda = df.groupby(['Category', 'Returned']).agg(
    OrderCount=('OrderID', 'count'),
    TotalSales=('NetSales', 'sum')
)

return_eda.round(2)
```
